```python
import pandas as pd
from load_data import distorcao,enem_media,escola_por_1000,ideb,populacao_df,taxa_alfabetizacao,taxa_aprovacao_abandono
```

```python
df_distorcao = distorcao()
df_enem = enem_media()
df_escola = escola_por_1000()
df_ideb = ideb()
df_populacao = populacao_df()
df_alfabetizacao = taxa_alfabetizacao()
df_aprovacao_abandono = taxa_aprovacao_abandono()
```

```python
resultado = df_distorcao
resultado = resultado.merge(df_enem, on='Cidades', how='outer')
resultado = resultado.merge(df_escola, on='Cidades', how='outer')
resultado = resultado.merge(df_ideb, on='Cidades', how='outer')
resultado = resultado.merge(df_populacao, on='Cidades', how='outer')
resultado = resultado.merge(df_alfabetizacao, on='Cidades', how='outer')
resultado = resultado.merge(df_aprovacao_abandono, on='Cidades', how='outer')
resultado
resultado.to_csv("merge_final.csv", index=False)
```
